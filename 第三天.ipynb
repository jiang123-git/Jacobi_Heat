1.矩阵乘法 C = A @ B

```python
import numpy as np
def matmul_manual(A,B):
    return A@B
A=np.array([[1,2],
            [2,3]])
B=np.array([[2,3],
            [3,4]])
print(matmul_manual(A,B))
```

2.矩阵的转置

```python
import numpy as np
def transpose_manual(A):
    return A.T
A=np.array([[1,2],[3,4]])
print(transpose_manual(A))
```

3.返回n×n的单位矩阵

```python
import numpy as np
def identity(n):
    return np.ones((n,n))
print(identity(3))
```

4.迹（对角线元素之和）

```python
import numpy as np
def trace_manual(A):
    return np.sum(np.diag(A))
A=np.array([[1,2],[2,3]])
print(trace_manual(A))
```

5. Frobenius 范数 $\|A\|_F = \sqrt{\sum_{ij} a_{ij}^2}$

```python
import numpy as np
def fro_norm(A):
    return np.sqrt(np.sum(A**2))
A=np.array([[1,2,2]])
print(fro_norm(A))
```

6. 行和范数（无穷范数）$\|A\|_\infty = \max_i \sum_j |a_{ij}|$

```python
import numpy as np
def norm_inf(A):
    return np.max(np.sum(np.abs(A),axis=1))
A=np.array([[1,2,3],[4,5,6]])
print(norm_inf(A))
```

7.上三角矩阵的行列式 = 对角线元素之积。

```python
import numpy as np
def det_upper(U):
    return np.prod(np.diag(U))
A=np.array([[1,2],
            [0,4]])
print(det_upper(A))
```

9.解线性方程组 A x = b（克拉默法则或消元公式都行）

```python
import numpy as np
def solve_A(A,b):
    return np.linalg.solve(A,b)
A=np.array([[1,2],[3,4]])
B=np.array([5,11])
print(solve_A(A,B))
print(A@solve_A(A,B))
```

10.是否严格对角占优(每一行对角线元素绝对值>其余元素绝对值的和)

```python
import numpy as np
def is_diag_dominant(A):
    return np.all(np.abs(np.diag(A)))>(np.sum(np.abs(A),axis=1)-np.abs(np.diag(A)))
A=np.array([[1,2],[2,3]])
print(is_diag_dominant(A))
```

11.判断 Q 的**列向量**是否标准正交

```python
import numpy as np
def check_orthogonal(Q):
    n=Q.shape[1]
    I_approx=Q.T@Q
    I_true=np.eye(n)
    return np.allclose(I_approx,I_true)
Q,R=np.linalg.qr(np.random.rand(3,3))
print(check_orthogonal(Q))
```

12.无穷范数条件数$\mathrm{cond}_\infty(A) = \|A\|_\infty \cdot \|A^{-1}\|_\infty$

```python
import numpy as np

def inf_condition_number(A):
    # 计算 A 的无穷范数
    norm_A = np.linalg.norm(A, np.inf)
    
    # 计算 A 的逆矩阵
    try:
        A_inv = np.linalg.inv(A)
    except np.linalg.LinAlgError:
        return float('inf')  # 奇异矩阵，条件数为无穷大
        
    # 计算 A_inv 的无穷范数
    norm_A_inv = np.linalg.norm(A_inv, np.inf)
    
    # 返回条件数
    return norm_A * norm_A_inv

# 测试示例
A = np.array([[1, 2], [3, 4]])
cond_inf = inf_condition_number(A)
print(f"A的无穷范数条件数: {cond_inf}")
```
