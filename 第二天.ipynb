1.返回 float 数组 [0, 2, 4, ..., 2n-2]，长度正好是 n。
    n = 0 时返回长度为 0 的空数组。

```python
import numpy as np
def get_array(n):
   A=np.arange(0,2*n-2,2).astype(float)
   return A
print(get_array(9))
```

2.返回把 x 的【第一个元素和最后一个元素都去掉】之后得到的数组

```python
import numpy as np
def get_array(n):
   A=np.arange(0,2*n-2,2).astype(float)   
   return A
B=get_array(9)
C=B[1:-1]
print(C)
```

3.返回 x 里所有【偶数】，保持原来的先后顺序

```python
import numpy as np
def evens(x):
    return x[x%2==0]
test_array=np.array([-2,4,6,5])
result=evens(test_array)
print(result)
```

4.返回 x 里【严格大于 0】的元素的个数

```python
import numpy as np
def A(x):
    return int((x>0).sum())
test_array=np.array([1,2,-3,-6])
result_array=A(test_array)
print(result_array)
```

5.返回 x 减去它自己的平均值之后的数组（去均值）

```python
import numpy as np
def center(x):
    return x-x.mean()
A=np.array([1,2,3])
print(center(A))
```

6.标准化：(x - 均值) / 标准差。标准差用【总体标准差】（除以 n，不是 n-1）

```python
import numpy as np
def normalize(x):
    return (x-x.mean())/x.std()
A=np.array([1,2,3])
print(normalize(A))
```

7.M 是 (n, m) 的二维数组，返回【每一行的最大值】，形状 (n,)

```python
import numpy as np
def row_max(M):
    return M.max(axis=1)
A=np.array([[1,2,3],
            [4,5,6]])
print(row_max(A))
```

8.M 是 (n, m) 的二维数组，返回【每一列的和】，形状 (m,)

```python
import numpy as np
def col_sum(M):
    return M.sum(axis=0)
A=np.array([[1,2,3],
            [4,5,6]])
print(col_sum(A))
```

9.a 长 p，b 长 q，返回形状 (p, q) 的矩阵，其中第 i 行第 j 列 = a[i] + b[j]

```python
import numpy as np
def outer_sum(a,b):
    return a[:,None]+b[None,:]
A=np.array([1,2])
B=np.array([10,20,30])
print(outer_sum(A,B))
```

12.返回 max|a - b|：两个等长数组逐元素相减、取绝对值、再取最大值，得到**一个数**。

```python
import numpy as np
def max_abs_diff(a,b):
    return float(np.abs(a-b).max())
A=np.array([1,2,3])
B=np.array([1,2,3.5])
print(max_abs_diff(A,B))
```

```python
import numpy as np

# 列主元：寻找第k列，k行往下绝对值最大的行号
def pivot_row(A, k):
    A_a = A[k:, k]
    abs_A_a = np.abs(A_a)
    argmax_abs_A_a = np.argmax(abs_A_a)
    return int(k + argmax_abs_A_a)

# 前向消元，返回上三角矩阵U和新右端c
def forward_elim(A, b):
    A = np.array(A, dtype=float, copy=True)
    b = np.array(b, dtype=float, copy=True)
    n = len(b)
    
    for k in range(n - 1):
        p = pivot_row(A, k)
        if p != k:
            # 交换k行 p行
            A[[k, p], :] = A[[p, k], :]
            b[[k, p]] = b[[p, k]]
        
        if abs(A[k, k]) < 1e-12:
            raise ValueError("奇异阵，不可解")
        
        for i in range(k + 1, n):
            factor = A[i, k] / A[k, k]
            A[i, k:] = A[i, k:] - factor * A[k, k:]
            b[i] = b[i] - factor * b[k]
    return A, b

# 回代求解上三角 Ux = c
def back_substitution(U, c):
    n = len(c)
    x = np.zeros(n)
    for i in range(n - 1, -1, -1):
        if abs(U[i,i]) < 1e-12:
            raise ValueError("奇异阵")
        sum_part = U[i, i+1:] @ x[i+1:]
        x[i] = (c[i] - sum_part) / U[i, i]
    return x

# 高斯求解总函数
def gauss_solve(A,b):
    U,c = forward_elim(A,b)  # ✅ 传b，不是C！
    x = back_substitution(U,c)
    return x

# 测试数据
A = np.array([[2, 1, -1],
              [-3, -1, 2],
              [-2, 1, 2]], dtype=float)
b = np.array([8, -11, -3], dtype=float)

x = gauss_solve(A,b)
print("自己写的高斯消元结果：",x)
print("numpy内置求解结果：",np.linalg.solve(A,b))
```
