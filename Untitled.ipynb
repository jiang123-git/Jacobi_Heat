```python
import numpy as np
import matplotlib.pyplot as plt

def jacobi_heat_2d(N, tol=1e-4, max_iter=1000):
    # 1. 初始化：创建 N x N 的温度矩阵 T，全部设为 0
    T = np.zeros((N, N))
    
    # 2. 设置边界条件
    T[0, :] = 0      # 上边界 0度
    T[N-1, :] = 0    # 下边界 0度
    T[:, 0] = 100    # 左边界 100度
    T[:, N-1] = 0    # 右边界 0度
    
    # 记录误差历史，用于后续画图
    error_history = []
    
    # 3. 迭代过程
    for k in range(max_iter):
        # 创建新矩阵 T_new，全设为 0
        T_new = np.zeros((N, N))
        error = 0.0
        
        # 遍历内部所有点（不去碰边界）
        # 注意：Python 中 range(1, N-1) 对应伪代码中的 1 到 N-2
        for i in range(1, N-1):
            for j in range(1, N-1):
                # 核心公式：上下左右四个邻居的平均值
                T_new[i, j] = (T[i-1, j] + T[i+1, j] + T[i, j-1] + T[i, j+1]) / 4.0
                # 计算误差
                error = max(error, abs(T_new[i, j] - T[i, j]))
        
        # 边界条件也要赋给 T_new，防止边界丢失
        T_new[0, :] = 0
        T_new[N-1, :] = 0
        T_new[:, 0] = 100
        T_new[:, N-1] = 0
        
        # 记录本轮误差
        error_history.append(error)
        
        # 判断收敛
        if error < tol:
            print(f"在第 {k+1} 次迭代时收敛，误差 = {error:.6e}")
            break
        
        # 更新旧值，准备下一轮迭代
        T = T_new.copy() # 使用 copy() 确保不改变原来 T 的引用
    
    return T, error_history

# ================= 运行与验证 =================
# 设置网格尺寸 N=50，容差 1e-4，最大迭代 2000 次
N = 50
T, errors = jacobi_heat_2d(N, tol=1e-4, max_iter=2000)

# 画热力图
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
# 注意：T 是 [行, 列] 即 [y, x]。
# imshow 默认按行显示，需要转置或者用 origin='lower' 来匹配实际的 x-y 坐标
plt.imshow(T, cmap='hot', origin='lower', extent=[0, 1, 0, 1])
plt.colorbar(label='温度 (℃)')
plt.title(f'钢板温度分布热力图 (N={N})')
plt.xlabel('x')
plt.ylabel('y')

# 画误差收敛曲线
plt.subplot(1, 2, 2)
plt.semilogy(errors, 'b-') # 用对数坐标更清晰地看误差下降
plt.xlabel('迭代次数')
plt.ylabel('最大误差 (对数坐标)')
plt.title('Jacobi 迭代误差下降曲线')
plt.grid(True, which="both", ls="--")

plt.tight_layout()
plt.show()
```
