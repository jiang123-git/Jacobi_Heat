1.列主元——在第 k 列里，从第 k 行往下（就是子列 A[k:, k]），
    找出绝对值最大的那个元素，返回它的**绝对行号**（一个 Python int）。
    并列时返回行号最小的那个。k 之前的行不许看（它们已经消好了）。

```python
import numpy as np
def pivot_row(A,k):
    k=1
    sub_col=A[k:,k]  #从第k行开始，在第k列
    abs_sub_col=np.abs(sub_col)
    relative_index=np.argmax(abs_sub_col)  #返回位置，max是返回值
    return int(k+relative_index)
A=np.array([[1,2,3],
            [4,5,6],
            [7,-10,9],
            [10,10,12]])
print(f"截取到的子列A[{k}:,{k}]={sub_col}")
print(pivot_row(A,k))
```

2.带列主元的前向消元，返回二元组 (U, c)

```python
import numpy as np
# 第一题：列主元
def pivot_row(A, k):
    sub_col = A[k:, k]  # 从第k行开始，在第k列
    abs_sub_col = np.abs(sub_col)
    relative_index = np.argmax(abs_sub_col) # 返回位置
    return int(k + relative_index)
# 第二题：带列主元的前向消元
def forward_elim(A, b):
    """
    带列主元的前向消元，返回二元组 (U, c)
    """
    # 1. 拷贝数据，转为浮点型
    A = np.array(A, dtype=float, copy=True) #不加数据会乱
    b = np.array(b, dtype=float, copy=True)
    n = len(b)
    for k in range(n - 1):
        p = pivot_row(A, k)
        if p != k:
            A[[k, p]] = A[[p, k]]
            b[[k, p]] = b[[p, k]]
        if A[k, k] == 0:
            raise ValueError("矩阵奇异，无法继续消元")
        for i in range(k + 1, n):
            factor = A[i, k] / A[k, k]
            A[i, k+1:] = A[i, k+1:] - factor * A[k, k+1:]
            b[i] = b[i] - factor * b[k]
            A[i, k] = 0.0
    return A, b

# 测试运行区域（验证你的代码）
if __name__ == "__main__":
    A_test = np.array([[2, 1, -1],
                       [-3, -1, 2],
                       [-2, 1, 2]])
    b_test = np.array([8, -11, -3])
    U, c = forward_elim(A_test, b_test)
    print("原始矩阵 A:\n", A_test)
    print("原始向量 b:", b_test)
    print("\n====== 运行结果 ======")
    print("上三角矩阵 U:\n", U)
    print("变换后的向量 c:", c)
    print("\nU 的下三角部分（应全是 0）:\n", np.tril(U, -1)) #不保留主对角线，保留下方，其他全为0
```

3.回代。U 是上三角矩阵，解 U @ x = c，返回解向量 x。

```python
import numpy as np

def back_substitution(U, c):
    n = len(c)  # 获取方程组的维数 
    x = np.zeros(n)     # 提示3：先开好存放结果的数组，后面直接填值，避免动态扩容
    for i in range(n - 1, -1, -1):   # 提示1：从最后一行往上走 (n-1, n-2, ..., 0)
        if U[i, i] == 0:
            raise ValueError(f"矩阵奇异：第 {i} 行主对角线元素为 0，无法求解")
        sum_part = U[i, i+1:] @ x[i+1:]
        x[i] = (c[i] - sum_part) / U[i, i]
    return x
# 测试数据
A_test = np.array([[2, 1, -1],
                   [-3, -1, 2],
                   [-2, 1, 2]])
b_test = np.array([8, -11, -3])
U, c = forward_elim(A_test, b_test)
x = back_substitution(U, c)
print("解得向量 x =", x)
x_true = np.linalg.solve(A_test, b_test) # 验证一下：用 numpy 自带的求解器对比
print("标准答案 x =", x_true)
```

4.前代。L 是**单位下三角**（对角元为 1），解 L @ x = c，返回解向量 x。

```python
import numpy as np

def forward_sub(L, c):
    n = len(c)
    x = np.zeros(n)
    for i in range(n):
        if L[i, i] == 0:
            raise ValueError(f"矩阵奇异：第 {i} 行主对角线元素为 0，无法求解")
        sum_part = L[i, :i] @ x[:i]
        x[i] = (c[i] - sum_part) / L[i, i]    
    return x
# --- 测试代码 ---
L_test = np.array([[1, 0, 0],
                   [2, 1, 0],
                   [3, 4, 1]], dtype=float)
c_test = np.array([5, 8, 20], dtype=float)

x_sol = forward_sub(L_test, c_test)
print("解得向量 x =", x_sol)
# 用 numpy 自带求解器验证
x_true = np.linalg.solve(L_test, c_test)
print("标准答案 x =", x_true)
```

5.高斯消元（列主元）解 A @ x = b，返回解向量 x。

```python
import numpy as np
def pivot_row(A,k):
    A_a=A[k:,k]
    abs_A_a=np.abs(A_a)
    argmax_abs_A_a=np.argmax(abs_A_a)
    return int(k+argmax_abs_A_a)
def forward_elim(A,b):
    A=np.array(A,dtype=float,copy=True)
    b=np.array(b,dtype=float,copy=True)
    n=len(b)
    for k in range(n-1):
        p=pivot_row(A,k)
        if p!=k:
            A[[k,p],:]=A[[p,k],:]
            b[[k,p]]=b[[p,k]]
        if A[k,k]==0:
            raise ValueError("奇异阵，不可")
        for i in range(k+1,n):
            factor=A[i,k]/A[k,k]
            A[i,k:]=A[i,k:]-factor*A[k,k:]
            b[i]=b[i]-factor*b[k]
    return A,b
def back_substitution(U,c):
    n=len(c)
    x=np.zeros(n)
    for i in range(n-1,-1,-1):
        if U[i,i]==0:
            raise ValueError("奇异阵")
        sum_part=U[i,i+1:]@x[i+1:]
        x[i]=(c[i]-sum_part)/U[i,i]
    return x

# --- 构造测试数据 ---

def gauss_solve(A,b):
    U,c=forward_elim(A,b)
    x=back_substitution(U,c)
    return x
A= np.array([[2, 1, -1],
                   [-3, -1, 2],
                   [-2, 1, 2]], dtype=float)
b= np.array([8, -11, -3], dtype=float)
print(x)
print(np.linalg.solve(A,b))
```

6.用消元法求行列式（返回一个 float）。

```python
import numpy as np
def det_by_elim(A):
    A = np.array(A, dtype=float)
    n = A.shape[0]
    if n == 1:
        return float(A[0, 0])
    sign = 1.0  # 用于记录行交换的次数，偶数次不变号，奇数次变号
    for j in range(n - 1):
        max_idx = np.argmax(np.abs(A[j:, j])) + j
        if A[max_idx, j] == 0:
            return 0.0
        if max_idx != j:
            A[[j, max_idx], :] = A[[max_idx, j], :]
            sign *= -1
        for i in range(j + 1, n):
            factor = A[i, j] / A[j, j]
            A[i, j:] -= factor * A[j, j:]
    diagonal_product = np.prod(np.diag(A))
    return float(sign * diagonal_product)
A = [
    [0, 2, 1],
    [1, 3, 2],
    [2, 1, 4]
]
det_by_elim(A)
```

7.残差的无穷范数，即 max|b - A @ x|（返回一个 float）。
    算完方程组先看这个数：到 1e-12 量级才算解对了。

```python
import numpy as np
def residual_inf(A,x,b):
    diff=b-A@x
    return float(np.max(np.abs(diff)))
A=np.array([[2,1],
           [1,3]])
x=np.array([1,1])
b=np.array([3,4])
print(residual_inf(A,x,b))
b_bug=[3.0000000000001,4]
print(residual_inf(A,x,b_bug))
```

8.返回 n×n 的 Hilbert 矩阵，元素 H[i, j] = 1 / (i + j + 1)。

```python
import numpy as np
def hilbert(n):
    return 1.0/(np.arange(n)[:,None]+np.arange(n)[None,:]+1) 
           #arange(n)生成0，1，...,n-1。
           #[:,None]自动分成n行1列，[None,:]自动分成1行n列
print(hilbert(3))
```

9.相对误差 = max|x_num - x_true| / max|x_true|（返回一个 float）。

```python
import numpy as np
def relative_error(x_num,x_true):
    x_num=np.array(x_num,dtype=float)
    x_true=np.array(x_true,dtype=float)
    numerator=np.max(np.abs(x_num-x_true))
    denominator=np.max(np.abs(x_true))
    return float(numerator/denominator)
x_num=[1.0001,2.0002,3.0001]
x_true=[1.0,2.0,3.0]
print("相对误差:",relative_error(x_num,x_true))
```

10.给定 n 个点 (xs[i], ys[i])，求过这 n 个点的 n-1 次多项式
        p(x) = c[0] + c[1] x + c[2] x^2 + ... + c[n-1] x^(n-1)
    的系数 c，返回长度 n 的数组，**顺序是从常数项到最高次项**。

```python
# 你可以参考这个修正版的 forward_elim
def forward_elim(A, b):
    A = np.array(A, dtype=float)
    b = np.array(b, dtype=float)
    n = len(b)
    
    for k in range(n - 1):
        # 选主元，防止除以 0
        max_idx = np.argmax(np.abs(A[k:, k])) + k
        if A[max_idx, k] == 0:
            raise ValueError("矩阵奇异，无唯一解")
        
        # 交换行
        if max_idx != k:
            A[[k, max_idx], :] = A[[max_idx, k], :]
            b[k], b[max_idx] = b[max_idx], b[k]
            
        # 消元
        for i in range(k + 1, n):
            factor = A[i, k] / A[k, k]
            A[i, k:] -= factor * A[k, k:]
            b[i] -= factor * b[k]
            
    return A, b  # 只返回两个变量

def gauss_solve(A, b):
    U, c = forward_elim(A, b)
    # 这里假设你的回代函数名为 back_substitution
    x = back_substitution(U, c)
    return x
# 测试：求过 (1, 3), (2, 5), (3, 9) 的二次多项式
# 理论上多项式是 y = x^2 + x + 1，系数应为 [1, 1, 1]
xs_test = [1.0, 2.0, 3.0]
ys_test = [3.0, 5.0, 9.0]

c = poly_interp_coeffs(xs_test, ys_test)
print("计算得到的系数:", c)
# 预期输出接近: [1. 1. 1.]
```
